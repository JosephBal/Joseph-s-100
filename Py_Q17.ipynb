Question: 
Write a program that computes the net amount of a bank account based a transaction log from console input. The transaction log format is shown as following: D 100 W 200

D means deposit while W means withdrawal. 

Suppose the following input is supplied to the program: D 300 D 300 W 200 D 100 
Then, the output should be: 500

Hints: In case of input data being supplied to the question, it should be assumed to be a console input.



```python
# Create transaction history

account_balance = 0
while True:
    x = input()
    if not x:
        break
    value = x.split(" ")
    operation = value[0]
    amount = int(value[1])
    if operation == "D":
        account_balance +=amount
    elif operation == "W":
        account_balance -=amount
else:
    pass

# Output remaining balance

print(account_balance)
```
