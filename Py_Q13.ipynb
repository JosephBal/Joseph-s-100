Question: Write a program that accepts a sentence and calculate the number of letters and digits. Suppose the following input is supplied to the program: hello world! 123 Then, the output should be: LETTERS 10 DIGITS 3

Hints: In case of input data being supplied to the question, it should be assumed to be a console input.



```python
# Input sentence

x = input("Write a sentence:")

# Calculte # of letters & digits seperately

y = {"letters":0, "digits":0}
for count in x:
    if count.isdigit():                      # .isdigit() will count the digits in a string
        y["digits"]+=1
    elif count.isalpha():                    # .isalpha() will count the letters in a string
        y["letters"]+=1
    else:
        pass

# Print Output
print(" ")
print("Letters:",y["letters"],"    ","Numbers:",y["digits"]) # recheck the code count seems to be off
```
