Question: Write a program that computes the value of a+aa+aaa+aaaa with a given digit as the value of a. Suppose the following input is supplied to the program: 9 Then, the output should be: 11106

Hints: In case of input data being supplied to the question, it should be assumed to be a console input.



```python
# Input Number and assign to a

n = input("insert a single-digit here (0-9):")

# Formula and expression

x1 = int(("%s" % n))
x2 = int(("%s%s" % (n,n)))
x3 = int(("%s%s%s" % (n,n,n)))
x4 = int(("%s%s%s%s" % (n,n,n,n)))
y = x1+x2+x3+x4

# Print Output
print("")
print(y)
```
